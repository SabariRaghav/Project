```python
inventory = {"Rice":{"price": 60, "qty": 3},
             "Sugar":{"price": 45, "qty": 10},
             "Oil":{"price": 120, "qty": 6},
             "Milk":{"price": 30, "qty": 25}}

def restock(product,quantity):
    if product in inventory:
        quantity=quantity+5
        print(product,"product restocked successfully")
    else:
        print("Product not found")
        
def sell_product(product,quantity):
    if product not in inventory:
        print("Product not found")
        return
    
    avaliable_quantity=inventory[product]["qty"]
        
    user_required_quantity=input("Enter the quantity:")
    if quantity>avaliable_quantity:
        raise("InsufficientStockError,Only{avaliable_quantity} items are avaliable")
        
    inventory[product]["qty"]-=quantity
    print(quantity,product,"Sold")
        
def low_stock_report():
    
    print("Low stock product(qty<5)")
    
    for product in inventory:
        if inventory[product]["qty"]<5:
            print(product,"qty:",inventory[product]["qty"])
            
print("Initial Inventory:")
print(inventory)

restock("Rice",9)

try:
    sell_product("Sugar",5)
    sell_product("Oil",6)
except InsufficientStockError as Error:
    print("Error",Error)

low_stock_report()

print("Final Inventory")
print(inventory)   
```
