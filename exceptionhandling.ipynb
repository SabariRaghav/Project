exception handling allow you to handle errors gracefully and take corrective action without stopping the execution of the program 
exceptions include intendation,value,attribute,name,index,key,type,zero division errors.
try:
    a=b 
except:
    print("Variable has not been assigned)



```python
try:
    a=b
except:
    print("Variable is not defined")
```

```python
try:
    a=b
except NameError as ex:
    print(ex)
    
```

```python
try:
    result=1/0
except ZeroDivisionError as ex:
    print(ex)
```

```python
try:
    num=int(input("enter a number:"))
    result=10/num
except ValueError as ex:
    print(ex)
except ZeroDivisionError as ex1:
    print(ex1)
else:
    print(result)
```
