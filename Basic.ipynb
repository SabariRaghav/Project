```python
#keyword arguments:
def sample(**kwargs):
    for i,j in kwargs.items():
        print(f"{i}:{j}")
    
```

```python
sample(name="raghav",age=22,country="india")
```

```python
def sample2(*args,**kwargs):
    for i in args:
        print(i)
    for j,k in kwargs.items():
        print(f"{j}:{k}")
```

Filter() iterator of an element. It return which element is true. It is used filter out from a list based on condition

```python
lst=[12,34,56,79]
def even(num):
    if num%2==0:
        return True
    
```

```python
list(filter (even,lst))
```

```python
list(filter(lambda num:num>50,lst))
```

reduce() places a function cumulative to the elements of a list and return single final value. It process a elements step by step and combining two elements at a time until only one result remains.



```python
from functools import reduce
a=["apple","orange","mango"]
r=reduce(lambda x,y:x+ "" +y,a)
print(r)
```

```python
import operator
lst=[12,45,78]
print(reduce(operator.mul,lst)) 
```

```python
print(reduce(operator.add,a)) 
```

```python
from itertools import accumulate
from operator import add
a=[1,2,3]
b=accumulate(a,add)
print(list(b))
```
