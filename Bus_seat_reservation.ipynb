```python
seats=[None]*40

def book_seat(seat_number,name,destination):
    if seat_number<1 or seat_number>40:
        print("Invalid seat number")
        return
    
    if seats[seat_number]is None:
        seats[seat_number-1]=(name,destination)
        print("Seat",seat_number,"Book Successfully")
    else:
        print("Seat already Booked")
        
def cancel_booking(seat_number):
    if seat_number<1 or seat_number>40:
        print("Invalid seat number")
        return
    
    if seats[seat_number] is not None:
        seats[seat_number-1]=None
        print("Booking cancelled for seat",seat_number)
    else:
        print("Seat is already empty")
        
def print_manifest():
    manifest=[]
    for seat in seats:
        if seat is not None:
            manifest.append(seat)
        
    manifest.sort(key=lambda passenger: passenger[1])
    print("\nPassenger Manifested (sorted by destination):")
    
    for passenger in manifest:
        print("Name:",passenger[0],"| Destination:",passenger[1])
        
book_seat(2,"Raghav","Chennai")
book_seat(3,"Rishi","Chennai")
book_seat(5,"Ashwin","Chennai")
book_seat(7,"Vijay","Banglore")

cancel_booking(7)

print_manifest()
    
```
