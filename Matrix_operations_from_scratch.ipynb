```python
A = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

B = [
    [9, 8, 7],
    [6, 5, 4],
    [3, 2, 1]
]

def validate_matrix(matrix):
    if len(matrix)==0:
        raise ValueError("Matrix cannot be empty")
    
    column_length=len(matrix[0])
    
    for row in matrix:
        if len(row)!=column_length:
            raise ValueError("Invalid matrix: Rows have different size")
        
def add_matrix(A,B):
    
    validate_matrix(A)
    validate_matrix(B)
    
    if len(A)!=len(B) or len(A[0])!=len(B[0]):
        raise ValueError("Marix addition requires same dimensions")
    
    result=[]
    
    for i in range(len(A)):
        row=[]
        for j in range(len(A[0])):
            row.append(A[i][j]+B[i][j])
        result.append(row)
    return result

def transpose_matrix(matrix):
    
    validate_matrix(matrix)
    
    row=len(matrix)
    col=len(matrix[0])
    
    result=[]
    
    for j in range(col):
        new_row=[]
        
        for i in range(row):
            new_row.append(matrix[i][j])
        result.append(new_row)
        
    return result

def multiply_matrix(A,B):
    
    validate_matrix(A)
    validate_matrix(B)
    
    if len(A[0])!=len(B):
        raise ValueError("Matrix multiplication is not possible,Colums of A must be equal to rows of B")
    
    result=[]
    
    for i in range(len(A)):
        row=[]
        
        for j in range(len(B[0])):
            total=0
            
            for k in range(len(B)):
                total+=A[i][k]*B[k][j]
            
            row.append(total)
        result.append(row)
        
    return(result)

print("Matrix A:")
for row in A:
    print(row)
    
print("\nMatrix B")
for row in B:
    print(row)
    
print("\nMartix addition:")
addition_result=add_matrix(A,B)
for row in addition_result:
    print(row)
    
print("\nTranspose of Matrix A:")
transpose_result=transpose_matrix(A)
for row in transpose_result:
    print(row)
    
print("\nMatrix multiplication:")
multiplication_result=multiply_matrix(A,B)
for row in multiplication_result:
    print(row)    
        
    
    

                    
        
        
        
        
        
        
        
        
        
        
        
```
