```python
form_a=["a@x.com","b@x.com","c@x.com"]
form_b=["c@x.com","d@x.com","e@x.com"]

set_a=set(form_a)
set_b=set(form_b)

duplicates=set_a.intersection(set_b)

unique_a=set_a.difference(set_b)
unique_b=set_b.difference(set_a)

total_unique=set_a.union(set_b)

print("duplicates:",duplicates)
print("unique to form a:",unique_a)
print("unique to form b:",unique_b)
print("Total unique:",total_unique)
```
