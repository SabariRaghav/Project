```python
contacts={"raghav":"8300470970",
          "rishi":"8754356574",
          "denesh":"1234567890",
          "harish":"0987654321",
          "manoj":"9876543210"}

def add_contacts(name,phone):
    if name in contacts:
        print("Name aready exist")
    else:
        contacts[name]=phone
        print(name, "added successfully")
        
def delete_contact(name):
    if name in contacts:
        del contacts[name]
        print("contact deleted")
    else:
        print("Contact not found")
    
def search_contacts(name):
    if name in contacts:
        print("phone number of",name,"is",contacts[name])
    else:
        print("Contact not found")
        
def update_contact(name,new_phoneno):
    if name in contacts:
        contacts[name]=new_phoneno
        print(name,"updated successfully")
    else:
        print("contacts not found")

print("initial contacts")
print(contacts)

print("\nAdding Contacts:")
add_contacts("vijay","910034762") 

print("\nSearching contacts")
search_contacts("Rishi")

print("\nUpdate contact")
update_contact("raghav","9363007924")

print("\nDeleting contacts")
delete_contact("denesh")   

print("\nSearching missing contact")
search_contacts("harish")
        
print("\nFinal contacts")
print(contacts)
```
