```python
import matplotlib.pyplot as plt

```

```python
x=[1,2,3]
y=[4,5,8]

plt.plot(x,y)
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Line Plot")
plt.show()
```

```python
x=[1,2,5]
y=[4,5,6]

plt.plot(x,y,color="Red",linestyle="--",marker="o",linewidth=3,markerize=9)
plt.grid(True)
```
