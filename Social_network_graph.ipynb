```python
network = {
    "Alice": {"Bob", "Raj"},
    "Bob": {"Alice", "Priya"},
    "Raj": {"Alice", "Priya"},
    "Priya": {"Bob", "Raj"}
}

def add_friendship(person1,person2):
    if person1 not in network:
        network[person1]=set()
    if person2 not in network:
        network[person2]=set()
        
    network[person1].add(person2)
    network[person2].add(person1)
    
    print(person1,"and",person2, "are now friends")
    
def remove_friendship(person1,person2):
    if person1 in network and person2 in network[person1]:
        
        network[person1].remove(person2)
        network[person2].remove(person1)
        
        print(person1,"and",person2,"are no longer friends")
    
    else:
        print("Friendship does not exist")
        
def mutual_friendship(person1,person2):
    if person1 not in network or person2 not in network:
        print("User not found")
        return
    
    mutual=network[person1] & network[person2]
    print("Mutual friends of",person1,"and",person2,":")
    
    print(mutual)
    
def suggest_friends(person):
    if person not in network:
        print("User not found")
        return
    
    suggestions=set()

    for friends in network[person]:
        for friends_of_friends in network[friends]:
            if(friends_of_friends != person and friends_of_friends not in network[person]):
            
                suggestions.add(friends_of_friends)
    
    print("Friends suggestion for",person,":")
    print(suggestions)
    
print("Initial Network")
print(network)

add_friendship("Alice","Priya")

remove_friendship("Bob","Priya")

mutual_friendship("Alice","Priya")

suggest_friends("Bob")

print("\nFinal Network")
print(network)
```
