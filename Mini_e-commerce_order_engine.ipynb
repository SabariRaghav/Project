```python
inventory={
    "Laptop":10,"Mouse":25,"Keyboard":15,"Monitor":8,"Headphones":20
}

orders=[
    {
        "order_id":101,
        "customer":"Raghav",
        "items":[
            {"product":"Laptop","qty":1,"price":50000},
            {"product":"Mouse","qty":2,"price":700}
        ],
        
        "status":"Shipped"
    },
    
    {
        "order_id":102,
        "customer":"Rishi",
        "items":[
            {"product":"Keyboard","qty":1,"price":1500},
            {"product":"Mouse","qty":4,"price":700}
        ],
        
        "status":"Pending"
    },
    
    {
        "order_id":103,
        "customer":"Ashwin",
        "items":[
            {"product":"Monitor","qty":1,"price":15000},
            {"product":"Headphone","qty":3,"price":300}
        ],
        
        "status":"Shipped"
    },
    
    {
        "order_id":104,
        "customer":"Vijay",
        "items":[
            {"product":"Monitor","qty":1,"price":15000},
            {"product":"Mouse","qty":5,"price":700}
        ],
        
        "status":"Pending"
    },
]

def calculate_total_revenue(order_list):
    total_revenue=0
    for order in order_list:
        if order["status"]!="Cancelled":
            for item in order["items"]:
                total_revenue+=(item["qty"]*item["price"])
    return total_revenue

def revenue_per_customer(order_list):
    revenue={}
    for order in order_list:
        if order["status"]!="Cancelled":
            customer=order["customer"]
            
            if customer not in revenue:
                revenue[customer]=0
            for item in order["items"]:
                revenue[customer]+=(item["qty"]*item["price"])
    return revenue 

def most_ordered_product(order_list):
    product_count={}
    for order in order_list:
        if order["status"]!="Cancelled":
            for item in order["items"]:
                product=item["product"]
                qty=item["qty"]
                if product not in product_count:
                    product_count[product]=0
                product_count[product]+=qty
    top_product=None
    highest_qty=0
    
    for product,qty in product_count.items():
        
        if qty>highest_qty:
            highest_qty=qty
            top_product=product
    return top_product,highest_qty

def pending_order(order_list):
    pending=[]
    for order in order_list:
        if order["status"]=="Pending":
            pending.append(order)
    return pending

def cancel_order(order_list,inventory,order_id):
    for order in order_list:
        if order["order_id"]==order_id:
            if order["status"]=="Cancelled":
                print("Order is already Cancelled")
                return
            order_status="Cancelled"
            for item in order["items"]:
                product=item["product"]
                qty=item["qty"]
                
                inventory[product]+=qty
            print("Order",order_id,"Got Cancelled")
            return
    print("Order_id not found")
    
print("----Total Revenue----")
print(calculate_total_revenue(orders))

print("\n----Revenue per customer----")
customer_revenue=revenue_per_customer(orders)
for customer, revenue in customer_revenue.items():
    print(customer,"->",revenue)
        
print("\n----Most Ordered Product----")
product,qty=most_ordered_product(orders)
print(product,"ordered",qty,"times")

print("\n----Pending orders----")
pending=pending_order(orders)
for order in pending:
    print("Order_id:",order["order_id"],"|Customer:",order["customer"])
    
print("\n----Cancel Order----")
cancel_order(orders,inventory,104)

print("\n----Update Inventory----")
for product,qty in inventory.items():
    print(product,"->",qty)
    
print("\n----Updated order status----")
for order in orders:
    print(order["order_id"],"->",order["status"])
            
                      















```
