```python
students={
    "Raghav":{
        "Maths":88,
        "Physics":90,
        "Computer Science":99
    },
    
    "Rishi":{
        "Maths":90,
        "Physics":97,
        "Computer Science":100
    },
     "Ashwin":{
        "Maths":92,
        "Physics":87,
        "Computer Science":84
    },
      "Vijay":{
        "Maths":78,
        "Physics":76,
        "Computer Science":89
    }
}

subject_topper={}

student_average={}

failed_students=[]

class_total=0
total_subjects=0

for student,subjects in students.items():
    
    student_total=0
    failed=False
    for subject,score in subjects.items():
        student_total+=score
        class_total+=score
        total_subjects+=1
        
        if subject not in subject_topper:
            subject_topper[subject]=(student,score)
        else:
            topper_name,topper_score=subject_topper[subject]
            
            if score>topper_score:
                subject_topper[subject]=(student,score)
        
        if score<35:
            failed=True

    average=student_total/len(subjects)
    
    student_average[student]=average
    
    if failed:
        failed_students.append(students)

overall_average=class_total/total_subjects


print("----SUBJECT TOPPERS----")
for subject,topper_data in subject_topper.items():
    name,marks=topper_data
    print(subject,"->",name,"(",marks,")")
    
print("\n----STUDENT AVERAGE----")
for student,average in student_average.items():
    print(student,"->",round(average,2))
    
print("\n----STUDENTS FAILED IN ANY SUBJECTS----")
print(failed_students)

print("\n----OVERALL CLASS AVERAGE----")
print(round(overall_average,2)) 
```
